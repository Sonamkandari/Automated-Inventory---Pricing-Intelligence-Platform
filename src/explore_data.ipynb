```python
print(df.groupby("Category")["Demand"].sum().sort_values(ascending=False))

# now we have our first useful business insight
# Groceries has by far the highest total demand 

```

```python
# Now lets find out which individual product has the highest total demand
print(df.groupby("Product ID")["Demand"].sum().sort_values(ascending=False).head(10))
```

```python
# Using this we P0009 product has the highest demand , the total demand of this product and Highest-demand category
# also we get that which product has the lowest demand
```

```python
# Now Let's Investigate P0009's inventory 

print(df[df["Product ID"] == "P0009"]["Inventory Level"].mean())
```

```python
# P0009 has the highest demand, but is its inventory enough to support that demand?
print(df[df["Product ID"] == "P0009"]["Units Sold"].mean())
```

```python
# Check how Much P0009 is being ordered on average
print(df[df["Product ID"] == "P0009"]["Units Ordered"].mean())
```

```python
# Now check the P0009's average demand
print(df[df["Product ID"] == "P0009"]["Price"].mean())
```

```python
# let's check P0009's average competitor price.
print(df[df["Product ID"] == "P0009"]["Competitor Pricing"].mean())
```

```python
# calculate the exact price difference 
print(df[df["Product ID"] == "P0009"]["Competitor Pricing"].mean() - df[df["Product ID"] =="P0009"]["Price"].mean())
```
